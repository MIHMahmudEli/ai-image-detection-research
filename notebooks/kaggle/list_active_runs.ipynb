# List Active Runs
Cross-run status dashboard via HuggingFace API.

Shows all runs in the checkpoint repo, their status, best metrics, and flags any manifest SHA-256 mismatches.

**Run this LOCALLY** (not on Kaggle) — it reads your local `.env` for credentials.

```python
# Cell 1: Clone repo
import subprocess, sys
from pathlib import Path

REPO_DIR = Path("/kaggle/working/mfft_repo")
if not REPO_DIR.exists():
    subprocess.run(["git", "clone", "https://github.com/MIHMahmudEli/ai-image-detection-research.git", str(REPO_DIR)], check=True)
sys.path.insert(0, str(REPO_DIR))

subprocess.run(["pip", "install", "-q", "huggingface_hub", "python-dotenv"], check=False)
print("Ready.")
```
