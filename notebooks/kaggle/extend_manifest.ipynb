# Extend Manifest
Manually add newly uploaded Kaggle dataset shards to the frozen manifest.

**When to use:** You've uploaded new dataset shards to Kaggle and want to include them in training.

**What it does:**
1. Downloads current manifest from HF
2. Discovers new shards not already in `shards_used`
3. Splits new images with same ratios (stratified, seed=42)
4. Archives old version as `split_manifest_vN.json`
5. Uploads new version to `manifest/split_manifest.json`

**Old runs are NOT invalidated** — they keep their `manifest_sha256` and can still be resumed.

```python
# Cell 1: Clone repo & install deps
import subprocess, sys, shutil
from pathlib import Path

REPO_DIR = Path("/kaggle/working/mfft_repo")
if REPO_DIR.exists():
    shutil.rmtree(REPO_DIR)
subprocess.run(["git", "clone", "https://github.com/MIHMahmudEli/ai-image-detection-research.git", str(REPO_DIR)], check=True)
sys.path.insert(0, str(REPO_DIR))

subprocess.run(["pip", "install", "-q", "huggingface_hub", "scikit-learn", "python-dotenv"], check=False)
print("Ready.")
```

```python
# Cell 2: Verify HF token
try:
    from kaggle_secrets import UserSecretsClient
    token = UserSecretsClient().get_secret("HF_TOKEN")
    print(f"HF Token: {token[:8]}...")
except Exception as e:
    print(f"ERROR: HF_TOKEN not found: {e}")
```

## After extending
